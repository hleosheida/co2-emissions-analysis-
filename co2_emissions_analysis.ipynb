```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv")
df.head()
```

```python
print(df.shape)
df[["country", "year", "co2", "population"]].describe()
```

```python
print(df.shape)
df[["country", "year", "co2", "population"]].describe()
```

```python
sweden = df[df["country"] == "Sweden"]
sweden.plot(x="year", y="co2", title="Sweden CO2 emissions over time")
```

```python
import matplotlib.pyplot as plt

countries = ["Sweden", "Germany", "India", "United States"]
subset = df[df["country"].isin(countries)]

for c in countries:
    data = subset[subset["country"] == c]
    plt.plot(data["year"], data["co2_per_capita"], label=c)

plt.title("CO2 emissions per person over time")
plt.xlabel("Year")
plt.ylabel("Tonnes of CO2 per person")
plt.legend()
plt.show()
```

In 2022, china had the highest total CO2 emissions, followed by United States and India. The top 10 includes both large industrial economies and fast-growing countries. Total emissions depend on population size, so a country like India ranks high in total but much lower per person. This is why it’s important to look at both total and per-person emissions.”
